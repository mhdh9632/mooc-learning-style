```python
import pandas as pd
import matplotlib.pyplot as plt

print("Loading raw data for exploration...")
log_train = pd.read_csv("data/raw/train/train/log_train.csv")
truth_train = pd.read_csv("data/raw/train/train/truth_train.csv", header=None, names=["student_id", "dropout"])

# --- 1. Class Imbalance (Dropouts vs Retained) ---
print("\n--- Student Dropout Distribution ---")
dropout_counts = truth_train['dropout'].value_counts()
print(dropout_counts)

plt.figure(figsize=(6, 4))
dropout_counts.plot(kind='bar', color=['red', 'green'])
plt.title('Student Dropout (1) vs Stay (0)')
plt.xticks(rotation=0)
plt.ylabel('Number of Students')
plt.show()

# --- 2. Most Common Actions ---
print("\n--- Top 10 Most Frequent Events ---")
top_events = log_train['event'].value_counts().head(10)
print(top_events)

plt.figure(figsize=(8, 4))
top_events.plot(kind='barh', color='steelblue')
plt.title('Top 10 User Events')
plt.gca().invert_yaxis()
plt.xlabel('Frequency')
plt.show()
```
