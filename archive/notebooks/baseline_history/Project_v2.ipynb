```python
import pandas as pd
import numpy as np

print("Loading datasets...")

log_train = pd.read_csv("train/train/log_train.csv")
log_test = pd.read_csv("test/test/log_test.csv")
object_df = pd.read_csv("object.csv")

log_train['time'] = pd.to_datetime(log_train['time'])
log_test['time'] = pd.to_datetime(log_test['time'])


# -----------------------------
# LOAD TRUTH LABELS
# -----------------------------

truth_df = pd.read_csv("train/train/truth_train.csv", header=None)
truth_df.columns = ["student_id","dropout"]

print("Train Logs shape:", log_train.shape)
print("Test Logs shape:", log_test.shape)
print("Objects shape:", object_df.shape)


# -----------------------------
# MERGE CATEGORY INFORMATION
# -----------------------------

log_train = log_train.merge(
    object_df[['module_id','category']],
    how='left',
    left_on='object',
    right_on='module_id'
)

log_test = log_test.merge(
    object_df[['module_id','category']],
    how='left',
    left_on='object',
    right_on='module_id'
)

#changed module id to object, should be reversed if not right(if vocabulary size doesnt increase)
# log_train = log_train.merge(
#     object_df[['object','category']],
#     how='left',
#     on='object'
# )

# log_test = log_test.merge(
#     object_df[['object','category']],
#     how='left',
#     on='object'
# )

print("NaN % BEFORE fillna:",
      log_train['category'].isna().mean())
print("NaN % BEFORE fillna:",
      log_test['category'].isna().mean())

log_train['category'] = log_train['category'].fillna("unknown")
log_test['category'] = log_test['category'].fillna("unknown")


# -----------------------------
# SORT EVENTS
# -----------------------------

log_train = log_train.sort_values(['enrollment_id','time'])
log_test = log_test.sort_values(['enrollment_id','time'])


# -----------------------------
# COMPUTE TIME GAP
# -----------------------------

log_train['time_gap'] = (
    log_train.groupby('enrollment_id')['time']
    .diff()
    .dt.total_seconds()
    .fillna(0)
)

log_test['time_gap'] = (
    log_test.groupby('enrollment_id')['time']
    .diff()
    .dt.total_seconds()
    .fillna(0)
)

# log transform gaps (helps neural networks)
log_train['time_gap'] = np.log1p(log_train['time_gap'])
log_test['time_gap'] = np.log1p(log_test['time_gap'])


# -----------------------------
# CREATE EVENT TOKEN
# -----------------------------

log_train['token'] = (
    log_train['source'] + "_" +
    log_train['event'] + "_" +
    log_train['category']
)

log_test['token'] = (
    log_test['source'] + "_" +
    log_test['event'] + "_" +
    log_test['category']
)


# -----------------------------
# BUILD TRAIN SEQUENCES
# -----------------------------

train_tokens = (
    log_train.groupby("enrollment_id")["token"]
    .apply(list)
)

train_gaps = (
    log_train.groupby("enrollment_id")["time_gap"]
    .apply(list)
)

train_df = pd.DataFrame({
    "student_id": train_tokens.index,
    "tokens": train_tokens.values,
    "gaps": train_gaps.values
})


# -----------------------------
# MERGE TRAIN LABELS
# -----------------------------

train_df = train_df.merge(
    truth_df,
    how="inner",
    on="student_id"
)

print("\nTrain dataset shape:", train_df.shape)


# -----------------------------
# BUILD TEST SEQUENCES
# -----------------------------

test_tokens = (
    log_test.groupby("enrollment_id")["token"]
    .apply(list)
)

test_gaps = (
    log_test.groupby("enrollment_id")["time_gap"]
    .apply(list)
)

test_df = pd.DataFrame({
    "student_id": test_tokens.index,
    "tokens": test_tokens.values,
    "gaps": test_gaps.values
})

print("Test dataset shape:", test_df.shape)


# -----------------------------
# SHOW SAMPLE
# -----------------------------

sample = train_df.sample(1).iloc[0]

print("\nSample Student:", sample["student_id"])
print("Dropout Label:", sample["dropout"])

print("\nFirst 10 events:\n")

for i,(token,gap) in enumerate(zip(sample["tokens"][:10], sample["gaps"][:10])):
    print(i+1, token, "| gap =", gap)


# -----------------------------
# SAVE DATASETS
# -----------------------------

train_df.to_pickle("mooc_train_sequences.pkl")
test_df.to_pickle("mooc_test_sequences.pkl")

print("\nSaved:")
print("mooc_train_sequences.pkl")
print("mooc_test_sequences.pkl")
```

```python
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from torch.nn.utils.rnn import pack_padded_sequence
import pandas as pd
import numpy as np

print("Loading sequence datasets...")

train_df = pd.read_pickle("mooc_train_sequences.pkl")
test_df = pd.read_pickle("mooc_test_sequences.pkl")

print("Train size:", len(train_df))
print("Test size:", len(test_df))


# ------------------------------------------------
# SPLIT TOKEN INTO COMPONENTS
# ------------------------------------------------

def split_token(token):
    parts = token.split("_")
    if len(parts) == 3:
        return parts
    return ["unknown", "unknown", "unknown"]


# ------------------------------------------------
# BUILD VOCABS
# ------------------------------------------------

print("Building multi-feature vocabularies...")

source_set, event_set, category_set = set(), set(), set()

for seq in train_df["tokens"]:
    for token in seq:
        s, e, c = split_token(token)
        source_set.add(s)
        event_set.add(e)
        category_set.add(c)

source_vocab = {v:i+1 for i,v in enumerate(source_set)}
event_vocab = {v:i+1 for i,v in enumerate(event_set)}
category_vocab = {v:i+1 for i,v in enumerate(category_set)}

print("Source vocab:", len(source_vocab))
print("Event vocab:", len(event_vocab))
print("Category vocab:", len(category_vocab))


# ------------------------------------------------
# ENCODE MULTI-STREAM
# ------------------------------------------------

def encode_multi(seq):
    src, evt, cat = [], [], []
    for token in seq:
        s, e, c = split_token(token)
        src.append(source_vocab.get(s, 0))
        evt.append(event_vocab.get(e, 0))
        cat.append(category_vocab.get(c, 0))
    return src, evt, cat


train_encoded = train_df["tokens"].apply(encode_multi)
test_encoded = test_df["tokens"].apply(encode_multi)

train_df["src"] = train_encoded.apply(lambda x: x[0])
train_df["evt"] = train_encoded.apply(lambda x: x[1])
train_df["cat"] = train_encoded.apply(lambda x: x[2])

test_df["src"] = test_encoded.apply(lambda x: x[0])
test_df["evt"] = test_encoded.apply(lambda x: x[1])
test_df["cat"] = test_encoded.apply(lambda x: x[2])


# ------------------------------------------------
# SLIDING WINDOW
# ------------------------------------------------

MAX_LEN = 150
STRIDE = 75

def create_windows_multi(src, evt, cat, gaps):
    src_w, evt_w, cat_w, gap_w = [], [], [], []

    for i in range(0, len(src), STRIDE):
        s = src[i:i+MAX_LEN]
        e = evt[i:i+MAX_LEN]
        c = cat[i:i+MAX_LEN]
        g = gaps[i:i+MAX_LEN]

        if len(s) == 0:
            continue

        s += [0]*(MAX_LEN-len(s))
        e += [0]*(MAX_LEN-len(e))
        c += [0]*(MAX_LEN-len(c))
        g += [0]*(MAX_LEN-len(g))

        src_w.append(s)
        evt_w.append(e)
        cat_w.append(c)
        gap_w.append(g)

    return src_w, evt_w, cat_w, gap_w


# ------------------------------------------------
# BUILD TRAIN WINDOWS
# ------------------------------------------------

src_train, evt_train, cat_train, gap_train, y_train = [], [], [], [], []

for _, row in train_df.iterrows():
    s_w, e_w, c_w, g_w = create_windows_multi(
        row["src"], row["evt"], row["cat"], row["gaps"]
    )

    for s, e, c, g in zip(s_w, e_w, c_w, g_w):
        src_train.append(s)
        evt_train.append(e)
        cat_train.append(c)
        gap_train.append(g)
        y_train.append(row["dropout"])

src_train = np.array(src_train)
evt_train = np.array(evt_train)
cat_train = np.array(cat_train)
gap_train = np.array(gap_train)
y_train = np.array(y_train)

print("Windowed train samples:", len(src_train))


# ------------------------------------------------
# BUILD TEST WINDOWS
# ------------------------------------------------

src_test, evt_test, cat_test, gap_test, test_ids = [], [], [], [], []

for _, row in test_df.iterrows():
    s_w, e_w, c_w, g_w = create_windows_multi(
        row["src"], row["evt"], row["cat"], row["gaps"]
    )

    for s, e, c, g in zip(s_w, e_w, c_w, g_w):
        src_test.append(s)
        evt_test.append(e)
        cat_test.append(c)
        gap_test.append(g)
        test_ids.append(row["student_id"])

src_test = np.array(src_test)
evt_test = np.array(evt_test)
cat_test = np.array(cat_test)
gap_test = np.array(gap_test)


# ------------------------------------------------
# DATASET
# ------------------------------------------------

class SeqDataset(Dataset):

    def __init__(self, src, evt, cat, gaps, y=None):
        self.src = torch.LongTensor(src)
        self.evt = torch.LongTensor(evt)
        self.cat = torch.LongTensor(cat)
        self.gaps = torch.FloatTensor(gaps)
        self.y = torch.FloatTensor(y) if y is not None else None

    def __len__(self):
        return len(self.src)

    def __getitem__(self, idx):
        if self.y is None:
            return self.src[idx], self.evt[idx], self.cat[idx], self.gaps[idx]
        return self.src[idx], self.evt[idx], self.cat[idx], self.gaps[idx], self.y[idx]


train_loader = DataLoader(
    SeqDataset(src_train, evt_train, cat_train, gap_train, y_train),
    batch_size=128, shuffle=True
)

test_loader = DataLoader(
    SeqDataset(src_test, evt_test, cat_test, gap_test),
    batch_size=128
)


# ------------------------------------------------
# MODEL
# ------------------------------------------------

class BiLSTMModel(nn.Module):

    def __init__(self):
        super().__init__()

        self.source_emb = nn.Embedding(len(source_vocab)+1, 4, padding_idx=0)
        self.event_emb = nn.Embedding(len(event_vocab)+1, 8, padding_idx=0)
        self.category_emb = nn.Embedding(len(category_vocab)+1, 8, padding_idx=0)

        self.lstm = nn.LSTM(
            input_size=4+8+8+1,
            hidden_size=64,
            batch_first=True,
            bidirectional=True
        )

        #  ADD ATTENTION LAYER
        self.attention = nn.Linear(128, 1)

        self.fc = nn.Linear(128, 1)

    def forward(self, src, evt, cat, gaps, lengths):

        # embeddings
        s = self.source_emb(src)
        e = self.event_emb(evt)
        c = self.category_emb(cat)

        gaps = gaps.unsqueeze(-1)

        x = torch.cat([s, e, c, gaps], dim=2)

        # pack
        packed = pack_padded_sequence(
            x, lengths.cpu(), batch_first=True, enforce_sorted=False
        )

        packed_out, _ = self.lstm(packed)

        # 🔥 UNPACK to get all timesteps
        out, _ = torch.nn.utils.rnn.pad_packed_sequence(
            packed_out, batch_first=True
        )   # shape: (B, T, 128)

        # ------------------------------------------------
        # 🔥 SELF-ATTENTION STARTS HERE
        # ------------------------------------------------

        # compute attention scores
        attn_scores = self.attention(out).squeeze(-1)   # (B, T)

        # mask padding (VERY IMPORTANT)
        # get actual sequence length after unpacking
        seq_len = out.size(1)

        # trim mask
        mask = (src[:, :seq_len] != 0)

        attn_scores[~mask] = -1e9

        # softmax over time
        attn_weights = torch.softmax(attn_scores, dim=1)   # (B, T)

        # weighted sum
        attn_weights = attn_weights.unsqueeze(-1)          # (B, T, 1)
        context = torch.sum(out * attn_weights, dim=1)     # (B, 128)

        # ------------------------------------------------

        return torch.sigmoid(self.fc(context)).squeeze()


device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = BiLSTMModel().to(device)

print("Using device:", device)


# ------------------------------------------------
# TRAINING
# ------------------------------------------------

criterion = nn.BCELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)

for epoch in range(3):

    model.train()
    total_loss = 0

    for src, evt, cat, gaps, labels in train_loader:

        src, evt, cat = src.to(device), evt.to(device), cat.to(device)
        gaps, labels = gaps.to(device), labels.to(device)

        lengths = (src != 0).sum(dim=1)

        optimizer.zero_grad()

        outputs = model(src, evt, cat, gaps, lengths)

        loss = criterion(outputs, labels)

        loss.backward()
        optimizer.step()

        total_loss += loss.item()

    print(f"Epoch {epoch+1} Loss:", total_loss / len(train_loader))


# ------------------------------------------------
# PREDICTION + AGGREGATION
# ------------------------------------------------

model.eval()
window_preds = []

with torch.no_grad():

    for src, evt, cat, gaps in test_loader:

        src, evt, cat = src.to(device), evt.to(device), cat.to(device)
        gaps = gaps.to(device)

        lengths = (src != 0).sum(dim=1)

        outputs = model(src, evt, cat, gaps, lengths)

        window_preds.extend(outputs.cpu().numpy())


pred_df = pd.DataFrame({
    "student_id": test_ids,
    "pred": window_preds
})

final_preds = pred_df.groupby("student_id")["pred"].mean().reset_index()

final_preds.rename(columns={"pred": "dropout_probability"}, inplace=True)

final_preds.to_csv("mooc_test_predictions.csv", index=False)

print("\nSaved predictions: mooc_test_predictions.csv")
```
