```python
import pandas as pd
import numpy as np

print("Loading datasets...")

log_train = pd.read_csv("train/train/log_train.csv")
log_test = pd.read_csv("test/test/log_test.csv")
object_df = pd.read_csv("object.csv")

log_train['time'] = pd.to_datetime(log_train['time'])
log_test['time'] = pd.to_datetime(log_test['time'])


# -----------------------------
# LOAD TRUTH LABELS
# -----------------------------

truth_df = pd.read_csv("train/train/truth_train.csv", header=None)
truth_df.columns = ["student_id","dropout"]

print("Train Logs shape:", log_train.shape)
print("Test Logs shape:", log_test.shape)
print("Objects shape:", object_df.shape)


# -----------------------------
# MERGE CATEGORY INFORMATION
# -----------------------------

log_train = log_train.merge(
    object_df[['module_id','category']],
    how='left',
    left_on='object',
    right_on='module_id'
)

log_test = log_test.merge(
    object_df[['module_id','category']],
    how='left',
    left_on='object',
    right_on='module_id'
)

log_train['category'] = log_train['category'].fillna("unknown")
log_test['category'] = log_test['category'].fillna("unknown")


# -----------------------------
# SORT EVENTS
# -----------------------------

log_train = log_train.sort_values(['enrollment_id','time'])
log_test = log_test.sort_values(['enrollment_id','time'])


# -----------------------------
# COMPUTE TIME GAP
# -----------------------------

log_train['time_gap'] = (
    log_train.groupby('enrollment_id')['time']
    .diff()
    .dt.total_seconds()
    .fillna(0)
)

log_test['time_gap'] = (
    log_test.groupby('enrollment_id')['time']
    .diff()
    .dt.total_seconds()
    .fillna(0)
)

# log transform gaps (helps neural networks)
log_train['time_gap'] = np.log1p(log_train['time_gap'])
log_test['time_gap'] = np.log1p(log_test['time_gap'])


# -----------------------------
# CREATE EVENT TOKEN
# -----------------------------

log_train['token'] = (
    log_train['source'] + "_" +
    log_train['event'] + "_" +
    log_train['category']
)

log_test['token'] = (
    log_test['source'] + "_" +
    log_test['event'] + "_" +
    log_test['category']
)


# -----------------------------
# BUILD TRAIN SEQUENCES
# -----------------------------

train_tokens = (
    log_train.groupby("enrollment_id")["token"]
    .apply(list)
)

train_gaps = (
    log_train.groupby("enrollment_id")["time_gap"]
    .apply(list)
)

train_df = pd.DataFrame({
    "student_id": train_tokens.index,
    "tokens": train_tokens.values,
    "gaps": train_gaps.values
})


# -----------------------------
# MERGE TRAIN LABELS
# -----------------------------

train_df = train_df.merge(
    truth_df,
    how="inner",
    on="student_id"
)

print("\nTrain dataset shape:", train_df.shape)


# -----------------------------
# BUILD TEST SEQUENCES
# -----------------------------

test_tokens = (
    log_test.groupby("enrollment_id")["token"]
    .apply(list)
)

test_gaps = (
    log_test.groupby("enrollment_id")["time_gap"]
    .apply(list)
)

test_df = pd.DataFrame({
    "student_id": test_tokens.index,
    "tokens": test_tokens.values,
    "gaps": test_gaps.values
})

print("Test dataset shape:", test_df.shape)


# -----------------------------
# SHOW SAMPLE
# -----------------------------

sample = train_df.sample(1).iloc[0]

print("\nSample Student:", sample["student_id"])
print("Dropout Label:", sample["dropout"])

print("\nFirst 10 events:\n")

for i,(token,gap) in enumerate(zip(sample["tokens"][:10], sample["gaps"][:10])):
    print(i+1, token, "| gap =", gap)


# -----------------------------
# SAVE DATASETS
# -----------------------------

train_df.to_pickle("mooc_train_sequences.pkl")
test_df.to_pickle("mooc_test_sequences.pkl")

print("\nSaved:")
print("mooc_train_sequences.pkl")
print("mooc_test_sequences.pkl")
```

```python
import pandas as pd

print("Loading data...")

log_train = pd.read_csv("train/train/log_train.csv")
object_df = pd.read_csv("object.csv")

log_train['time'] = pd.to_datetime(log_train['time'])

print("Logs shape:", log_train.shape)
print("Objects shape:", object_df.shape)

chapters = object_df[object_df['category'] == "chapter"].copy()

chapters = chapters.reset_index(drop=True)
chapters["chapter_num"] = chapters.index + 1

print("Total chapters:", len(chapters))

sections = object_df[object_df["category"] == "sequential"].copy()

sections["chapter_num"] = None
sections["section_num"] = None

for _, chapter in chapters.iterrows():

    if pd.isna(chapter["children"]):
        continue

    children = chapter["children"].split()

    for i, child in enumerate(children):

        sections.loc[
            sections["module_id"] == child,
            "chapter_num"
        ] = chapter["chapter_num"]

        sections.loc[
            sections["module_id"] == child,
            "section_num"
        ] = i + 1
object_df["chapter_num"] = None
object_df["section_num"] = None

for _, sec in sections.iterrows():

    if pd.isna(sec["children"]):
        continue

    children = sec["children"].split()

    for child in children:

        object_df.loc[
            object_df["module_id"] == child,
            "chapter_num"
        ] = sec["chapter_num"]

        object_df.loc[
            object_df["module_id"] == child,
            "section_num"
        ] = sec["section_num"]
object_df.loc[
    object_df["module_id"].isin(sections["module_id"]),
    ["chapter_num","section_num"]
] = sections[["chapter_num","section_num"]].values

log_train = log_train.merge(
    object_df[["module_id","category","chapter_num","section_num"]],
    how="left",
    left_on="object",
    right_on="module_id"
)

log_train["chapter_num"] = log_train["chapter_num"].fillna(0).astype(int)
log_train["section_num"] = log_train["section_num"].fillna(0).astype(int)

log_train = log_train.sort_values(["enrollment_id","time"])

log_train["time_gap"] = (
    log_train.groupby("enrollment_id")["time"]
    .diff()
    .dt.total_seconds()
    .fillna(0)
)

log_train["event_token"] = (
    log_train["source"] + "_" +
    log_train["event"] + "_ch" +
    log_train["chapter_num"].astype(str) +
    "_sec" +
    log_train["section_num"].astype(str)
)

log_train["event_token"] = (
    log_train["event_token"] +
    "|gap=" +
    log_train["time_gap"].astype(int).astype(str)
)

event_sequences = (
    log_train
    .groupby("enrollment_id")["event_token"]
    .agg(list)
    .reset_index()
)

event_sequences.columns = ["student_id","event_sequence"]

sample = event_sequences.iloc[0]

print("\nSample Student:", sample["student_id"])
print("\nFirst 20 events:\n")

for i, event in enumerate(sample["event_sequence"][:20]):
    print(i+1, ":", event)

```

```python
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
import pandas as pd
import numpy as np
from collections import Counter

print("Loading sequence datasets...")

train_df = pd.read_pickle("mooc_train_sequences.pkl")
test_df = pd.read_pickle("mooc_test_sequences.pkl")

print("Train size:", len(train_df))
print("Test size:", len(test_df))


# ------------------------------------------------
# BUILD VOCABULARY FROM TRAIN TOKENS ONLY
# ------------------------------------------------

print("Building vocabulary...")

all_tokens = [token for seq in train_df["tokens"] for token in seq]

token_counts = Counter(all_tokens)

vocab = {token:i+1 for i,(token,_) in enumerate(token_counts.items())}

PAD = 0
vocab_size = len(vocab) + 1

print("Vocabulary size:", vocab_size)


# ------------------------------------------------
# ENCODE TOKEN SEQUENCES
# ------------------------------------------------

def encode(seq):
    return [vocab.get(token,0) for token in seq]

train_df["encoded"] = train_df["tokens"].apply(encode)
test_df["encoded"] = test_df["tokens"].apply(encode)


# ------------------------------------------------
# PAD SEQUENCES
# ------------------------------------------------

MAX_LEN = 150

def pad(seq):
    if len(seq) > MAX_LEN:
        seq = seq[:MAX_LEN]
    return seq + [0]*(MAX_LEN-len(seq))


def pad_gap(seq):
    if len(seq) > MAX_LEN:
        seq = seq[:MAX_LEN]
    return seq + [0]*(MAX_LEN-len(seq))


X_train = np.array(train_df["encoded"].apply(pad).tolist())
gap_train = np.array(train_df["gaps"].apply(pad_gap).tolist())
y_train = train_df["dropout"].values

X_test = np.array(test_df["encoded"].apply(pad).tolist())
gap_test = np.array(test_df["gaps"].apply(pad_gap).tolist())


# ------------------------------------------------
# DATASET CLASS
# ------------------------------------------------

class SeqDataset(Dataset):

    def __init__(self,X,gaps,y=None):

        self.X = torch.LongTensor(X)
        self.gaps = torch.FloatTensor(gaps)

        if y is not None:
            self.y = torch.FloatTensor(y)
        else:
            self.y = None

    def __len__(self):
        return len(self.X)

    def __getitem__(self,idx):

        if self.y is None:
            return self.X[idx], self.gaps[idx]

        return self.X[idx], self.gaps[idx], self.y[idx]


train_dataset = SeqDataset(X_train,gap_train,y_train)
test_dataset = SeqDataset(X_test,gap_test)


train_loader = DataLoader(train_dataset,batch_size=128,shuffle=True)
test_loader = DataLoader(test_dataset,batch_size=128)


# ------------------------------------------------
# MODEL
# ------------------------------------------------

class BiLSTMModel(nn.Module):

    def __init__(self,vocab_size):

        super().__init__()

        # token embeddings
        self.embedding = nn.Embedding(vocab_size,32,padding_idx=0)

        # LSTM (changed from BiLSTM)
        self.lstm = nn.LSTM(
            input_size=33,        # 32 embedding + 1 time gap
            hidden_size=64,
            batch_first=True
        )

        # classifier
        self.fc = nn.Linear(64,1)


    def forward(self,x,gaps):

        emb = self.embedding(x)

        gaps = gaps.unsqueeze(-1)

        x = torch.cat([emb,gaps],dim=2)

        _,(h,_) = self.lstm(x)

        h = h[-1]

        x = self.fc(h)

        return torch.sigmoid(x).squeeze()


device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

model = BiLSTMModel(vocab_size).to(device)

print("Using device:",device)


# ------------------------------------------------
# TRAINING SETUP
# ------------------------------------------------

criterion = nn.BCELoss()

optimizer = torch.optim.Adam(model.parameters(),lr=0.001)


# ------------------------------------------------
# TRAIN MODEL
# ------------------------------------------------

EPOCHS = 3

for epoch in range(EPOCHS):

    model.train()

    total_loss = 0

    for tokens,gaps,labels in train_loader:

        tokens = tokens.to(device)
        gaps = gaps.to(device)
        labels = labels.to(device)

        optimizer.zero_grad()

        outputs = model(tokens,gaps)

        loss = criterion(outputs,labels)

        loss.backward()

        optimizer.step()

        total_loss += loss.item()

    print("Epoch",epoch+1,"Loss:",total_loss/len(train_loader))


# ------------------------------------------------
# SAVE MODEL
# ------------------------------------------------

torch.save({
    "model_state_dict": model.state_dict(),
    "vocab": vocab,
    "max_len": MAX_LEN
},"mooc_bilstm_model.pt")

print("Model saved as mooc_bilstm_model.pt")


# ------------------------------------------------
# GENERATE TEST PREDICTIONS
# ------------------------------------------------

print("\nGenerating predictions on test set...")

model.eval()

predictions = []

with torch.no_grad():

    for tokens,gaps in test_loader:

        tokens = tokens.to(device)
        gaps = gaps.to(device)

        outputs = model(tokens,gaps)

        predictions.extend(outputs.cpu().numpy())


test_df["dropout_probability"] = predictions

print("\nSample predictions:")
print(test_df.head())


# ------------------------------------------------
# SAVE TEST PREDICTIONS
# ------------------------------------------------

test_df[["student_id","dropout_probability"]].to_csv(
    "mooc_test_predictions.csv",
    index=False
)

print("\nSaved predictions: mooc_test_predictions.csv")
```

```python
from sklearn.metrics import accuracy_score, roc_auc_score, precision_score, recall_score, f1_score
import pandas as pd

print("Evaluating model on test set...")

model.eval()

all_probs = []

with torch.no_grad():

    for tokens, gaps in test_loader:

        tokens = tokens.to(device)
        gaps = gaps.to(device)

        outputs = model(tokens, gaps)

        all_probs.extend(outputs.cpu().numpy())


# ------------------------------------------------
# ADD PROBABILITIES TO TEST DATAFRAME
# ------------------------------------------------

test_df["dropout_probability"] = all_probs


# ------------------------------------------------
# LOAD TRUE LABELS
# ------------------------------------------------

truth_df = pd.read_csv("test/test/truth_test.csv", header=None)
truth_df.columns = ["student_id","dropout"]

eval_df = test_df.merge(truth_df, on="student_id")


# ------------------------------------------------
# BINARY PREDICTIONS
# ------------------------------------------------

eval_df["predicted_dropout"] = (eval_df["dropout_probability"] > 0.5).astype(int)


# ------------------------------------------------
# METRICS
# ------------------------------------------------

accuracy = accuracy_score(
    eval_df["dropout"],
    eval_df["predicted_dropout"]
)

precision = precision_score(
    eval_df["dropout"],
    eval_df["predicted_dropout"]
)

recall = recall_score(
    eval_df["dropout"],
    eval_df["predicted_dropout"]
)

f1 = f1_score(
    eval_df["dropout"],
    eval_df["predicted_dropout"]
)

auc = roc_auc_score(
    eval_df["dropout"],
    eval_df["dropout_probability"]
)


print("\nModel Evaluation Metrics\n")

print("Accuracy :", accuracy)
print("Precision:", precision)
print("Recall   :", recall)
print("F1 Score :", f1)
print("AUC      :", auc)


# ------------------------------------------------
# DECILE LIFT
# ------------------------------------------------

print("\nComputing Decile Lift...")

eval_df["decile"] = pd.qcut(
    eval_df["dropout_probability"],
    10,
    labels=False,
    duplicates="drop"
)

overall_rate = eval_df["dropout"].mean()

decile_table = eval_df.groupby("decile").agg(
    students=("dropout","count"),
    dropouts=("dropout","sum")
)

decile_table["dropout_rate"] = (
    decile_table["dropouts"] /
    decile_table["students"]
)

decile_table["lift"] = (
    decile_table["dropout_rate"] /
    overall_rate
)

decile_table = decile_table.sort_index(ascending=False)

print("\nDecile Lift Table\n")
print(decile_table)

top_decile_lift = decile_table.iloc[0]["lift"]

print("\nTop Decile Lift:", top_decile_lift)
```
